```python
import requests
import json
```

Création d'un grand rectangle couvrant la france // Request de l'API

```python
url = "https://apicarto.ign.fr/api/aoc/appellation-viticole"

geometry = {
    "type": "Point",
    "coordinates": [-0.56, 47.47]
}

params = {
    "geom": json.dumps(geometry),
    "apikey": "api_key_180382313",
    "source": "prd"
}

response = requests.post(url, params=params)

print("Status :", response.status_code)
print(response.text[:2000])
```
